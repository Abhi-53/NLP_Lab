# Experiment 1 - Data Analysis

# Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('employee_information_100.csv')

df.head()
```

```python
df.describe()
```

# Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import matplotlib.pyplot as plt

avg_salary_department = df.groupby('Department')['Salary'].mean()

plt.figure(figsize=(8, 5))
plt.bar(avg_salary_department.index, avg_salary_department.values)
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.title('Average Salary by Department')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Experiment 1.2: Display the number of employees working in each department.

```python
employee_count_department = df['Department'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(employee_count_department.index, employee_count_department.values)
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.title('Number of Employees in each Department')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Experiment 1.3: Display the percentage of male and female employees using a pie chart

```python
male_count = (df['Gender'] == 'Male').sum()
female_count = (df['Gender'] == 'Female').sum()

plt.figure(figsize=(6, 6))
plt.pie([male_count, female_count], labels = ['Male', 'Female'], autopct = '%1.1f%%', startangle = 90)
plt.title('Percentage of Male and Female Employees')
plt.show()
```

# Experiment 1.4: Draw a histogram showing salary distribution.

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Salary'], bins = 10, edgecolor = 'Black')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.title('Salary Distribution')
plt.tight_layout()
plt.show()
```

# Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
import seaborn as sns

plt.figure(figsize=(8, 5))

for department, data in df.groupby('Department'):
  plt.scatter(data['Experience_Years'], data['Salary'], label = department)

plt.xlabel('Experience')
plt.ylabel('Salary')
plt.title('Relationship between Experience and Salary')
plt.legend(title = 'Department')
plt.tight_layout()
plt.show()
```

# Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top_10_employees = df.nlargest(10, 'Salary')

top_10_employees
```

# Experiment 1.7: Find the highest salary in every department.

```python
highest_salary_department = df.groupby('Department')['Salary'].max()

highest_salary_department
```

# Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
avg_salary = df['Salary'].mean()
employees_greater_salary = df[df['Salary'] > avg_salary]

employees_greater_salary
```

# Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_experience_department = df.groupby('Department')['Experience_Years'].mean()

avg_experience_department
```

# Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Age'], bins = 10, edgecolor = 'Black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Employee Ages')
plt.tight_layout()
plt.show()
```
