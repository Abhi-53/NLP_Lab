Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:
1. Convert text to lowercase. (Also display how many uppercase?)
2. Remove punctuation.        (Also display how many different types of punctuations?)
3. Remove numbers.            (Also display how many numbers?)
4. Remove extra whitespace. (How much extra whitespace is present?)
5. Display the cleaned text.     


```python
import re
import string

with open('2.1_text_data.txt', 'r', encoding = 'utf-8') as files:
  text = files.read()

print('Original text:\n')
print(text)
```

```python
#1. Convert text to lowercase.
uppercase_count = sum(1 for char in text if char.isupper())
print('Number of character in upper case: ', uppercase_count)

cleaned_text = text.lower()
print('\nLowercase text:\n')
print(cleaned_text)
```

```python
# 2. Remove punctuation.
punctuation_types = set(char for char in text if char in string.punctuation)

print('Different types of punctuation: ', punctuation_types)
print('Number of different types of punctuation: ', len(punctuation_types))

cleaned_text = cleaned_text.translate(str.maketrans('', '', string.punctuation))
print('\nText after removing punctuation:\n')
print(cleaned_text)
```

```python
# 3. Remove numbers.
numbers = re.findall(r'\d+', text)
print('Count number: ', len(numbers))

cleaned_text = re.sub(r'\d+', '', cleaned_text)
print('\nText after removing numbers:\n')
print(cleaned_text)
```

```python
# 4. Remove extra whitespace.
extra_whitespace = re.findall(r'\s{2,}', text)
print('Count extra whitespace: ', len(extra_whitespace))

print('\nTotal extra whitespace present in text: ', sum(len(space) - 1 for space in extra_whitespace))

cleaned_text = re.sub(r'\s+', '', cleaned_text).strip()
print('\nText after removing extra whitespace:\n')
print(cleaned_text)
```

Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

```python
with open('2.2_tokenization_data.txt', 'r', encoding = 'utf-8') as files:
  text_2 = files.read()

print('Original text:\n')
print(text_2)
```

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

# Download necessary NLTK resources if not already downloaded
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

#Word tokenization
word_tokens = word_tokenize(text_2)
print('Word tokens:\n')
print(word_tokens)

#Sentence tokenization
sentence_tokens = sent_tokenize(text_2)
print('\nSentence tokens:\n')
for i, sentence in enumerate(sentence_tokens, start = 1):
  print(f"{i}. {sentence}")
```

Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """Natural Language Processing is a branch of Artificial Intelligence.
It helps computers understand and process human language.
In 2026, NLP is used in many applications such as chatbots, machine translation,
sentiment analysis, and text summarization."""

# Tokenize the text
tokens = word_tokenize(text)

# Get English stop words
stop_words = set(stopwords.words('english'))

# Find stop words present in the text
found_stop_words = [word for word in tokens if word.lower() in stop_words]

# Remove stop words
filtered_tokens = [
    word for word in tokens
    if word.lower() not in stop_words
]

# Display results
print("Original Tokens:")
print(tokens)

print("\nStop Words Found:")
print(found_stop_words)

print("\nFiltered Tokens:")
print(filtered_tokens)
```
