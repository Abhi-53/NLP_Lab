4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.
Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
Identify the top 10 most frequent terms and display their corresponding frequencies.

```python
import spacy
import csv
from collections import Counter

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Read input text
with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Process text
doc = nlp(text)

# Extract meaningful words
words = [
    token.text.lower()
    for token in doc
    if token.is_alpha and not token.is_stop
]

# Calculate term frequency
frequency = Counter(words)

# Save Term-Frequency data to CSV
with open("term_frequency_4_1.csv", "w",
          newline="", encoding="utf-8") as file:

    writer = csv.writer(file)
    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.most_common():
        writer.writerow([term, freq])

print("Term-frequency data saved to term_frequency_4_1.csv")

# Display top 10 terms
print("\nTop 10 Most Frequent Terms")
print("-" * 30)
print("Term\t\tFrequency")

for term, freq in frequency.most_common(10):
    print(f"{term:<15} {freq}")

# Named Entity Recognition
print("\nNamed Entities")
print("-" * 40)
print("Entity\t\tType")

for ent in doc.ents:
    print(f"{ent.text:<20} {ent.label_}")
```

4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import string
import csv

# Read input file
with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Convert to lowercase
text = text.lower()

# Remove punctuation
for char in string.punctuation:
    text = text.replace(char, " ")

# Tokenization
words = text.split()

# Dictionary for term frequency
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Sort according to frequency
sorted_frequency = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

# Save results in CSV
with open("term_frequency_4_2.csv", "w",
          newline="", encoding="utf-8") as file:

    writer = csv.writer(file)
    writer.writerow(["Term", "Frequency"])

    for term, freq in sorted_frequency:
        writer.writerow([term, freq])

print("Term-frequency data saved to term_frequency_4_2.csv")

# Display top 10 terms
print("\nTop 10 Most Frequent Terms")
print("-" * 30)
print("Term\t\tFrequency")

for term, freq in sorted_frequency[:10]:
    print(f"{term:<15} {freq}")
```

4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

1. Read multiple text documents.
2. Convert text to lowercase.
3. Remove punctuation.
4. Tokenize the documents.
5. Calculate TF for every term.
6. Calculate DF for every term.
7. Calculate IDF using the logarithmic formula.
8. Calculate TF-IDF for every term in every document.
9. Display Term, TF, DF, IDF, and TF-IDF at each step.
10. Identify the top 10 terms with the highest TF-IDF score for each document.


```python
import math
import string

# Step 1: Input documents

documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

# Preprocessing - Lowercase, remove punctuation and tokenize

processed_documents = []

for document in documents:

    # Convert to lowercase
    document = document.lower()

    # Remove punctuation
    for char in string.punctuation:
        document = document.replace(char, "")

    # Tokenization
    tokens = document.split()

    processed_documents.append(tokens)

print("PROCESSED DOCUMENTS")

for i, tokens in enumerate(processed_documents):
    print(f"Document {i+1}: {tokens}")


# Calculate TF

tf_values = []

for tokens in processed_documents:

    tf = {}
    total_terms = len(tokens)

    for term in tokens:
        tf[term] = tokens.count(term) / total_terms

    tf_values.append(tf)


# Calculate DF

df = {}

# Find all unique terms
all_terms = set()

for tokens in processed_documents:
    all_terms.update(tokens)

# Calculate Document Frequency
for term in all_terms:

    count = 0

    for tokens in processed_documents:
        if term in tokens:
            count += 1

    df[term] = count


# Calculate IDF

idf = {}

N = len(documents)

for term in all_terms:
    idf[term] = math.log(N / df[term])


# Calculate TF-IDF

tfidf_values = []

for tf in tf_values:

    tfidf = {}

    for term in tf:
        tfidf[term] = tf[term] * idf[term]

    tfidf_values.append(tfidf)


# Display - Term, TF, DF, IDF, TF-IDF

for i in range(len(documents)):

    print("\n" + "=" * 70)
    print(f"DOCUMENT {i+1}")
    print("=" * 70)

    print(
        f"{'Term':<15}"
        f"{'TF':<12}"
        f"{'DF':<8}"
        f"{'IDF':<12}"
        f"{'TF-IDF':<12}"
    )

    print("-" * 70)

    for term in tf_values[i]:

        print(
            f"{term:<15}"
            f"{tf_values[i][term]:<12.4f}"
            f"{df[term]:<8}"
            f"{idf[term]:<12.4f}"
            f"{tfidf_values[i][term]:<12.4f}"
        )


# Top 10 TF-IDF Terms

for i in range(len(documents)):

    sorted_terms = sorted(
        tfidf_values[i].items(),
        key=lambda x: x[1],
        reverse=True
    )

    print(f"\nTop 10 TF-IDF Terms - Document {i+1}")
    print("-" * 40)

    print(f"{'Term':<20}{'TF-IDF':<10}")

    for term, score in sorted_terms[:10]:
        print(f"{term:<20}{score:.4f}")
```
